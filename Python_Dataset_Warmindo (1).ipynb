**UNIT 5: MENGKONSTRUKSI DATA (J.62DMI00.009.1)**

```python
# Membuat dim_tanggal
min_date = df_clean['tanggal_transaksi'].min().floor('D')
max_date = df_clean['tanggal_transaksi'].max() + pd.offsets.MonthEnd(0)

date_range = pd.date_range(start=min_date, end=max_date, freq='D')

dim_tanggal = pd.DataFrame({'Date': date_range})
dim_tanggal['Tahun'] = dim_tanggal['Date'].dt.year
dim_tanggal['BulanNo'] = dim_tanggal['Date'].dt.month
dim_tanggal['Bulan'] = dim_tanggal['Date'].dt.strftime('%b')
dim_tanggal['TahunBulan'] = dim_tanggal['Date'].dt.strftime('%Y-%m')
dim_tanggal['HariNo'] = dim_tanggal['Date'].dt.weekday + 1 # Monday=1, Sunday=7

hari_map = {1: 'Senin', 2: 'Selasa', 3: 'Rabu', 4: 'Kamis', 5: 'Jumat', 6: 'Sabtu', 7: 'Minggu'}
dim_tanggal['Hari'] = dim_tanggal['HariNo'].map(hari_map)
```

```python
# Membuat dim_pelanggan
dim_pelanggan = df_clean.groupby('customer_id').agg(
    Frekuensi_Transaksi=('invoice_id', 'nunique'),
    Total_Belanja=('nilai_penjualan', 'sum')
).reset_index()
```

```python
print(dim_tanggal.head(5))
```

```python
# Menambahkan logika Segmentasi Pelanggan
def tentukan_segmen(row):
    if row['Frekuensi_Transaksi'] >= 8:
        return 'Loyal'
    elif row['Frekuensi_Transaksi'] >= 4:
        return 'Reguler'
    else:
        return 'Sesekali'

dim_pelanggan['Segmen'] = dim_pelanggan.apply(tentukan_segmen, axis=1)
```

```python
print(dim_pelanggan.head(5))
```

**UNIT 6: MENGINTEGRASIKAN DATA (J.62DMI00.011.1)**

```python
# Ekspor seluruh tabel dalam 1 file excel multi-sheet
output_file = "Dataset_Warmindo_Bersih.xlsx"

with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    df_clean.to_excel(writer, sheet_name='Fact_Penjualan', index=False)
    dim_tanggal.to_excel(writer, sheet_name='Dim_Tanggal', index=False)
    dim_pelanggan.to_excel(writer, sheet_name='Dim_Pelanggan', index=False)
```
